```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

```python
# 1. Read the raw CSVs directly into Pandas
df_gen = pd.read_csv('../data/raw/Plant_1_Generation_Data.csv')
df_weather = pd.read_csv('../data/raw/Plant_1_Weather_Sensor_Data.csv')

# 2. Parse the dates explicitly. 
# The generation data uses Day-Month-Year, while weather uses Year-Month-Day.
df_gen['DATE_TIME'] = pd.to_datetime(df_gen['DATE_TIME'], format='%d-%m-%Y %H:%M')
df_weather['DATE_TIME'] = pd.to_datetime(df_weather['DATE_TIME'], format='%Y-%m-%d %H:%M:%S')

# 3. Print the number of unique timestamps in each dataset
print(f"Unique generation timestamps: {df_gen['DATE_TIME'].nunique()}")
print(f"Unique weather timestamps: {df_weather['DATE_TIME'].nunique()}")
```

```python
# 1. Drop BOTH redundant columns from the weather DataFrame before merging
df_weather_clean = df_weather.drop(columns=['PLANT_ID', 'SOURCE_KEY'])

# 2. Inner merge to retain ONLY intervals where both sensors successfully recorded data
df_merged = pd.merge(df_gen, df_weather_clean, on='DATE_TIME', how='inner')

# 3. Rename the remaining SOURCE_KEY (which is now strictly the Inverter ID)
df_merged = df_merged.rename(columns={'SOURCE_KEY': 'inverter_id'})

# 4. Sort chronologically by inverter and timestamp
df_merged = df_merged.sort_values(by=['inverter_id', 'DATE_TIME']).reset_index(drop=True)

# 5. Inspect total rows and null count across all columns
print(f"Total merged rows: {len(df_merged)}")
print("\n--- Missing values per column ---")
print(df_merged.isnull().sum())
```

```python
# 1. Define the irradiance threshold for "daylight" (0.1 kW/m2)
min_irradiance = 0.1

# 2. Create a new column that flags True if there is a fault, and False if normal/night
# Condition: Irradiation is greater than 0.1 AND AC Power is exactly 0
df_merged['fault_zero_gen'] = (df_merged['IRRADIATION'] > min_irradiance) & (df_merged['AC_POWER'] == 0)

# 3. Count how many total intervals in our dataset have this fault
print("Total 15-minute intervals analyzed:", len(df_merged))
print("--- Zero-Generation Faults ---")
print(df_merged['fault_zero_gen'].value_counts())

# 4. Display a sample of the faulted rows to verify our logic worked
faults_only = df_merged[df_merged['fault_zero_gen'] == True]
print("\nSample of faulted rows:")
print(faults_only[['DATE_TIME', 'inverter_id', 'IRRADIATION', 'DC_POWER', 'AC_POWER']].head())
```

```python
faults_only['inverter_id'].nunique()
```

```python
# 1. Filter out the faults to find what "normal" looks like
healthy_inverters = df_merged[df_merged['fault_zero_gen'] == False]

# 2. Calculate the average AC power across all healthy inverters for each timestamp
expected_power = healthy_inverters.groupby('DATE_TIME')['AC_POWER'].mean().reset_index()
expected_power = expected_power.rename(columns={'AC_POWER': 'expected_ac_power'})

# 3. Merge this new 'expected_ac_power' baseline back into our main dataset
df_merged = pd.merge(df_merged, expected_power, on='DATE_TIME', how='left')

# 4. Calculate Power Lost (only for the faulted rows)
# If a fault occurred, the lost power is the expected power. Otherwise, 0.
df_merged['power_lost_W'] = df_merged.apply(
    lambda row: row['expected_ac_power'] if row['fault_zero_gen'] else 0, 
    axis=1
)

# 5. Convert Power (Watts) to Energy (kWh)
# AC_POWER in this dataset is in Watts. 
# We divide by 1000 for kW, and multiply by 0.25 (since 15 minutes is a quarter of an hour)
df_merged['energy_lost_kwh'] = (df_merged['power_lost_W'] / 1000) * 0.25

# 6. Summarize the total impact!
total_lost_kwh = df_merged['energy_lost_kwh'].sum()
print(f"Total estimated energy lost due to zero-generation faults: {total_lost_kwh:.2f} kWh")
```

```python
# 1. Define our target inverter and target date based on our previous findings
target_inverter = '1BY6WEcLGh8j5v7'
target_date = '2020-06-07'

# 2. Filter the dataset for this specific inverter on this specific day
# We use .dt.date to isolate just the Year-Month-Day portion of the timestamp
day_data = df_merged[(df_merged['inverter_id'] == target_inverter) & 
                     (df_merged['DATE_TIME'].dt.date.astype(str) == target_date)]

# 3. Initialize the plot with a professional, wide aspect ratio
plt.figure(figsize=(12, 6))

# 4. Plot the Expected Baseline (Dashed gray line)
plt.plot(day_data['DATE_TIME'], day_data['expected_ac_power'], 
         label='Expected AC Power (Plant Average)', color='gray', linestyle='--')

# 5. Plot the Actual Generation (Solid blue line)
plt.plot(day_data['DATE_TIME'], day_data['AC_POWER'], 
         label=f'Actual AC Power ({target_inverter})', color='#1f77b4', linewidth=2.5)

# 6. Highlight the faulted periods with red dots
faults = day_data[day_data['fault_zero_gen'] == True]
plt.scatter(faults['DATE_TIME'], faults['AC_POWER'], 
            color='red', label='Identified Fault (0W)', zorder=5, s=60)

# 7. Add professional formatting
plt.title(f'Inverter Outage Detection: {target_inverter} on {target_date}', fontsize=14, fontweight='bold')
plt.xlabel('Time of Day', fontsize=12)
plt.ylabel('AC Power', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)

# Clean up the x-axis to show hours/minutes nicely
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))
plt.xticks(rotation=45)
plt.tight_layout()

# 8. Save the figure to your main folder so you can upload it to GitHub!
plt.savefig('../fault_detection_plot.png', dpi=300)
plt.show()
```
