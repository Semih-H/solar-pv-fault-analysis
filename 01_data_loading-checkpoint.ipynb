```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the raw CSV files into pandas dataframes
df_gen = pd.read_csv('../data/raw/Plant_1_Generation_Data.csv')
df_weather = pd.read_csv('../data/raw/Plant_1_Weather_Sensor_Data.csv')

# Fix the date format mismatch so pandas recognizes them as actual Date/Time objects
# Generation uses Day-Month-Year, while weather uses Year-Month-Day
df_gen['DATE_TIME'] = pd.to_datetime(df_gen['DATE_TIME'], format='%d-%m-%Y %H:%M')
df_weather['DATE_TIME'] = pd.to_datetime(df_weather['DATE_TIME'], format='%Y-%m-%d %H:%M:%S')

# Check how many unique timestamps exist in both files 
# This tells me if there are missing intervals before I merge them together
print(f"Unique generation timestamps: {df_gen['DATE_TIME'].nunique()}")
print(f"Unique weather timestamps: {df_weather['DATE_TIME'].nunique()}")
```

```python
# Remove the weather station's ID columns to prevent duplicates during the merge
df_weather_clean = df_weather.drop(columns=['PLANT_ID', 'SOURCE_KEY'])

# Merge the datasets based on the timestamp. 
# 'inner' ensures we only keep rows where we have both generation AND weather data
df_merged = pd.merge(df_gen, df_weather_clean, on='DATE_TIME', how='inner')

# Rename the remaining SOURCE_KEY so anyone reading this knows it's the solar inverter
df_merged = df_merged.rename(columns={'SOURCE_KEY': 'inverter_id'})

# Organize the data chronologically for each individual inverter
df_merged = df_merged.sort_values(by=['inverter_id', 'DATE_TIME']).reset_index(drop=True)

# Do a final quality check to see how many rows I have and if there are any blank cells
print(f"Total merged rows: {len(df_merged)}")
print("\n--- Missing values per column ---")
print(df_merged.isnull().sum())
```

```python
# Set the minimum sunlight threshold. Anything above 0.1 means the sun is up.
min_irradiance = 0.1

# Flag the actual faults: Sun is shining, but the inverter is completely dead
df_merged['fault_zero_gen'] = (df_merged['IRRADIATION'] > min_irradiance) & (df_merged['AC_POWER'] == 0)

# Let's see how many intervals were perfectly normal versus how many were faults
print("Total 15-minute intervals analyzed:", len(df_merged))
print("--- Zero-Generation Faults ---")
print(df_merged['fault_zero_gen'].value_counts())

# Take a quick sample of just the bad intervals to make sure the logic makes sense
faults_only = df_merged[df_merged['fault_zero_gen'] == True]
print("\nSample of faulted rows:")
print(faults_only[['DATE_TIME', 'inverter_id', 'IRRADIATION', 'DC_POWER', 'AC_POWER']].head())
```

```python
# Count how many distinct inverters had at least one zero-generation fault
# This tells me if the failures are plant-wide or isolated to specific units
faults_only['inverter_id'].nunique()
```

```python
# Isolate the working inverters so we know what normal production looks like
healthy_inverters = df_merged[df_merged['fault_zero_gen'] == False]

# Find the average power produced by healthy units at every single 15-minute timestamp
expected_power = healthy_inverters.groupby('DATE_TIME')['AC_POWER'].mean().reset_index()
expected_power = expected_power.rename(columns={'AC_POWER': 'expected_ac_power'})

# Attach this expected baseline back to our main dataset
df_merged = pd.merge(df_merged, expected_power, on='DATE_TIME', how='left')

# If the inverter failed, the power we lost is that expected average. 
# If it worked normally, the loss is zero.
df_merged['power_lost_W'] = np.where(
    df_merged['fault_zero_gen'], df_merged['expected_ac_power'], 0
)


# Convert Watts to Kilowatt-hours (kWh). 
# Divide by 1000 for kW, and multiply by 0.25 (since a 15-minute interval is 1/4 of an hour)
df_merged['energy_lost_kwh'] = (df_merged['power_lost_W'] / 1000) * 0.25

# Add it all up to see the final operational impact
total_lost_kwh = df_merged['energy_lost_kwh'].sum()
print(f"Total estimated energy lost due to zero-generation faults: {total_lost_kwh:.2f} kWh")
```

```python
# Set up the exact inverter and date I want to zoom in on
target_inverter = '1BY6WEcLGh8j5v7'
target_date = '2020-06-07'

# 2. Filter the dataset for this specific inverter on this specific day
# We use .dt.date to isolate just the Year-Month-Day portion of the timestamp
day_data = df_merged[(df_merged['inverter_id'] == target_inverter) & 
                     (df_merged['DATE_TIME'].dt.date.astype(str) == target_date)]

# Set up a plot with a professional, wide aspect ratio for the chart
fig, ax = plt.subplots(figsize=(12, 6))

# 4. Plot the Expected Baseline (Dashed gray line)
ax.plot(day_data['DATE_TIME'], day_data['expected_ac_power'], 
        label='Expected AC Power (Plant Average)', color='gray', linestyle='--')

# 5. Plot the Actual Generation (Solid blue line)
ax.plot(day_data['DATE_TIME'], day_data['AC_POWER'], 
        label=f'Actual AC Power ({target_inverter})', color='#1f77b4', linewidth=2.5)

# 6. Highlight the faulted periods with red dots
faults = day_data[day_data['fault_zero_gen'] == True]
ax.scatter(faults['DATE_TIME'], faults['AC_POWER'], 
           color='red', label='Identified Fault (0W)', zorder=5, s=60)

# Add a bold title and label the axes so the chart is easy to read
ax.set_title(f'Inverter Outage Detection: {target_inverter} on {target_date}', fontsize=14, fontweight='bold')
ax.set_xlabel('Time of Day', fontsize=12)
ax.set_ylabel('AC Power', fontsize=12)
ax.legend()
ax.grid(True, alpha=0.3)

# Format the bottom axis to just show hours and minutes, tilted slightly to fit nicely
hourly_ticks = day_data.loc[day_data['DATE_TIME'].dt.minute == 0, 'DATE_TIME']
ax.set_xticks(hourly_ticks)
ax.set_xticklabels(hourly_ticks.dt.strftime('%H:%M'), rotation=45)

fig.tight_layout()

# 8. Save the figure to your main folder to use online, upload it to GitHub!
fig.savefig('../fault_detection_plot.png', dpi=300)
plt.show()
```

```python
# Save our final, enriched dataset into a new CSV file
# We set index=False so Pandas doesn't write the row numbers as a new column
df_merged.to_csv('../data/processed/plant_1_fault_analysis.csv', index=False)

print("Data exported successfully! Ready for Power BI.")
```
